```python
import numpy as np
import pandas as pd
df = pd.read_excel('/content/Book1.xlsx')
print("Columns in Book1.xlsx:", df.columns.tolist())
print(df.head())


target_column = 'AVERAGE

if target_column in df.columns:
    q1 = df[target_column].quantile(0.25)
    q3 = df[target_column].quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outliers = df[(df[target_column] < lower_limit) | (df[target_column] > upper_limit)]
    print(f"\nOutliers in {target_column}:")
    print(outliers)
else:
    print(f"\nColumn '{target_column}' not found in the DataFrame.")
```
